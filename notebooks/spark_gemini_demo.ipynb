# Spark + Gemini AI Integration Demo

This notebook demonstrates how to use Apache Spark with Google Gemini AI in your local home lab environment.

## Prerequisites
- Docker containers running (spark-master, spark-worker, jupyter)
- Google AI API key from https://aistudio.google.com/

## 1. Initialize Spark Session

````python
from pyspark.sql import SparkSession
import os

# Create Spark session
spark = SparkSession.builder \
    .appName("Gemini-Spark-Demo") \
    .master("spark://spark-master:7077") \
    .getOrCreate()

print(f"Spark Version: {spark.version}")
print(f"Spark Master: {spark.sparkContext.master}")
````

## 2. Create Sample Data

````python
# Create sample dataset
data = [
    ("Spark is great for big data processing", "technology", 1),
    ("Gemini AI provides powerful language models", "ai", 2),
    ("Home lab setups enable learning and experimentation", "education", 3),
    ("Docker makes deployment simple and consistent", "technology", 4),
    ("Open source tools democratize technology access", "community", 5)
]

columns = ["text", "category", "id"]
df = spark.createDataFrame(data, columns)

# Show the data
df.show(truncate=False)
````

## 3. Basic Spark Operations

````python
# Filter by category
print("Technology-related entries:")
df.filter(df.category == "technology").show(truncate=False)

# Count by category
print("\nCount by category:")
df.groupBy("category").count().show()
````

## 5. Save Results

````python
# Save to parquet format
output_path = "/home/jovyan/data/processed_data"
df.write.mode("overwrite").parquet(output_path)
print(f"Data saved to: {output_path}")
````

## 6. Clean Up

````python
# Stop Spark session
spark.stop()
print("Spark session stopped")
````
